### Prepare data for classification

```python
df = pd.read_csv('kidney_disease.csv')

df['classification'] = df['classification'].astype(str).str.strip().str.lower()
df = df[df['classification'].isin(['ckd', 'notckd'])].copy()
df['classification'] = df['classification'].map({'ckd': 1, 'notckd': 0}).astype(int)

y = df['classification']
X = df.drop(columns=['classification', 'id'], errors='ignore')

for col in X.select_dtypes(include=['object', 'string']).columns:
    X[col] = (
        X[col]
        .astype(str)
        .str.strip()
        .replace({'nan': 'Missing', 'None': 'Missing', '': 'Missing'})
    )

for col in X.select_dtypes(exclude=['object', 'string']).columns:
    X[col] = X[col].fillna(X[col].median())

X = pd.get_dummies(X, drop_first=True)

print('X shape:', X.shape)
print('y shape:', y.shape)
print('Target distribution:', y.value_counts().to_dict())
print('NaN check:', X.isna().sum().sum())
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)
```

### Classification Models

```python
from sklearn.metrics import accuracy_score

def classify(model, X, y):

    x_train, x_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.25,
        random_state=42,
        stratify=y
    )

    model.fit(x_train, y_train)

    y_pred = model.predict(x_test)

    accuracy = accuracy_score(
        y_test,
        y_pred
    ) * 100

    return accuracy
```

### Logestic Regression

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000
)

Logistic_acc = classify(
    model,
    X,
    y
)

print(
    "Logistic Regression Accuracy:",
    Logistic_acc
)
```

### Support Vector Machine

```python
from sklearn.svm import SVC

model = SVC(
    kernel='rbf',
    random_state=1
)

SVM_acc = classify(
    model,
    X,
    y
)

print(
    "SVM Accuracy:",
    SVM_acc
)
```

### Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    random_state=42
)

DecTree_acc = classify(
    model,
    X,
    y
)

print(
    "Decision Tree Accuracy:",
    DecTree_acc
)
```

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

RanFor_acc = classify(
    model,
    X,
    y
)

print(
    "Random Forest Accuracy:",
    RanFor_acc
)
```

### Accuracy Comparision

```python
Accuracy = [
    Logistic_acc,
    DecTree_acc,
    RanFor_acc,
    SVM_acc
]

models = [
    'Logistic Regression',
    'Decision Tree',
    'Random Forest',
    'Support Vector Machine'
]

accuracy_df = pd.DataFrame({
    'Model': models,
    'Accuracy': Accuracy
})

accuracy_df
```

### Accuracy Graph

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x='Model',
    y='Accuracy',
    data=accuracy_df
)

plt.title('Classification Model Accuracy Comparison')
plt.xlabel('Models')
plt.ylabel('Accuracy (%)')

plt.xticks(rotation=30)

plt.show()
```

### Confusion Matrix

```python
RanFor_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

x_train, x_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

RanFor_model.fit(
    x_train,
    y_train
)

y_pred = RanFor_model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(7,5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Not CKD', 'CKD'],
    yticklabels=['Not CKD', 'CKD']
)

plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')

plt.show()
```

### Classifiication Report

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred,
        target_names=['Not CKD', 'CKD']
    )
)
```

```python
final_accuracy = accuracy_score(
    y_test,
    y_pred
) * 100

print(
    "Random Forest Accuracy:",
    final_accuracy,
    "%"
)
```
